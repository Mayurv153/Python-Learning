# 📘 Python Functions - Day 1

A **function** is a reusable block of code that performs a specific task.  
Functions help us **organize code**, **avoid repetition**, and make programs **easier to read and maintain**.

### Syntax:
```python
def function_name(parameters):
    """docstring (optional)"""
    # body of the function
    return value  # optional
```

---

## 1️⃣ Basic Function (No Parameters, No Return)

The simplest type of function — it takes **no input** and **returns nothing**.  
It just executes the code inside it when called.

````python
# Defining a simple function with no parameters
def fun():
    print("hello mayur")  # This prints a greeting message

# Calling the function to execute it
fun()
````

## 2️⃣ Function with Parameters and Return Value

Functions can accept **parameters** (inputs) and **return** a result using the `return` keyword.  
Here we check if a number is **even or odd** using the modulo operator `%`.

````python
# Function that takes one parameter and returns a string
def evenOdd(x):
    if (x % 2 == 0):    # Check if x is divisible by 2
        return "Even"   # Return "Even" if remainder is 0
    else:
        return "Odd"    # Return "Odd" otherwise

# Testing with an even number
print(evenOdd(16))  # Output: Even

# Testing with an odd number
print(evenOdd(7))   # Output: Odd
````

## 3️⃣ Default Parameter Values

You can assign **default values** to function parameters.  
If the caller doesn't provide a value for that parameter, the **default is used** automatically.

```python
def myFun(x, y=50):  # y has a default value of 50
```

> 💡 **Note:** Default parameters must come **after** non-default parameters.

````python
# Function with a default parameter value for 'y'
def myFun(x, y=50):       # y defaults to 50 if not provided
    print("x: ", x)
    print("y: ", y)

# Calling with only one argument — y will use its default value (50)
myFun(10)
````

## 4️⃣ Keyword Arguments

When calling a function, you can pass arguments by **name** (keyword) instead of by **position**.  
This makes your code more **readable** and the **order doesn't matter**.

```python
student(fname='Geeks', lname='Practice')  # Using keyword arguments
student(lname='Practice', fname='Geeks')  # Order doesn't matter!
```

````python
# Function that accepts first name and last name
def student(fname, lname):
    print(fname, lname)

# Passing arguments using keyword names (order matches)
student(fname='Geeks', lname='Practice')

# Passing arguments in reverse order — still works because of keywords!
student(lname='Practice', fname='Geeks')
````

## 5️⃣ Practice: Keyword Arguments with Your Own Data

Same concept as above, but using **your own names** to practice calling functions with keyword arguments.

````python
# Redefining the student function (same as above)
def student(fname, lname):
    print(fname, lname)

# Calling with your own names using keyword arguments
student(fname='Mayur', lname='Vaidya')   # Output: Mayur Vaidya
student(fname='Ram', lname='Khajone')    # Output: Ram Khajone
````

## 6️⃣ Function with Multiple Parameters

Functions can accept **multiple parameters** to work with different pieces of data.  
Here, `nameage()` takes a **name** and an **age**, then prints a formatted introduction.

````python
# Function that takes name and age as parameters
def nameage(name, age):
    print('hi, I am', name)    # Print the name
    print('My age is ', age)   # Print the age

print("case-1:")  # Label for this test case

# Calling the function with positional arguments
nameage('mayur', 22)  # Output: hi, I am mayur | My age is 22
nameage('yesh', 21)   # Output: hi, I am yesh  | My age is 21
````

## 7️⃣ *args and **kwargs (Variable-Length Arguments)

Sometimes you don't know **how many arguments** a function will receive. Python provides:

| Syntax | Name | Purpose |
|--------|------|---------|
| `*args` | Non-keyword arguments | Accepts any number of **positional** arguments as a **tuple** |
| `**kwargs` | Keyword arguments | Accepts any number of **keyword** arguments as a **dictionary** |

### Example:
```python
def myFun(*args, **kwargs):
    # args  → ('Hey', 'Welcome')           → tuple
    # kwargs → {'first': 'Geeks', ...}      → dictionary
```

> 💡 **Tip:** `*args` must come **before** `**kwargs` in the function definition.

````python
# Function using *args (tuple) and **kwargs (dictionary)
def myFun(*args, **kwargs):
    # Loop through all positional arguments
    print("Non-Keyword Arguments (*args):")
    for arg in args:
        print(arg)  # Each positional arg is printed

    # Loop through all keyword arguments
    print("\nKeyword Arguments (**kwargs):")
    for key, value in kwargs.items():
        print(f"{key} == {value}")  # Print key-value pairs

# 'Hey' and 'Welcome' go into *args
# first, mid, last go into **kwargs
myFun('Hey', 'Welcome', first='Geeks', mid='for', last='Geeks')
````
